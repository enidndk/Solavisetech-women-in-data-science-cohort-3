```python
import numpy as np
import pandas as pd

# A list of dictionaries: each dictionary is one product
inventory = [
    {"product": "Notebook", "category": "Stationery",
     "price": 1500, "quantity": 40},
    {"product": "Pen", "category": "Stationery",
     "price": 200, "quantity": 120},
    {"product": "Backpack", "category": "Bags",
     "price": 12000, "quantity": 15},
    {"product": "Water Bottle", "category": "Drinkware",
     "price": 2500, "quantity": 0},
    {"product": "Laptop Sleeve", "category": "Bags",
     "price": 6500, "quantity": 8},
    {"product": "Highlighter Set", "category": "Stationery",
     "price": 1800, "quantity": None},
    {"product": "Coffee Mug", "category": "Drinkware",
     "price": 3000, "quantity": 22},
    {"product": "USB Flash Drive", "category": "Electronics",
     "price": 4500, "quantity": 12},
    {"product": "Earphones", "category": "Electronics",
     "price": 7000, "quantity": 5},
    {"product": "Pen", "category": "Stationery",
     "price": 200, "quantity": 120},
    {"product": "Desk Lamp", "category": "Electronics",
     "price": 9500, "quantity": 6},
]

# A simple dictionary of prices (used in Part 1)
prices = {"Notebook": 1500, "Pen": 200,
          "Backpack": 12000, "Water Bottle": 2500}

# Daily sales (in thousands of naira) for 14 days, 2 weeks (used in Part 3)
daily_sales = np.array([220, 250, 240, 230, 260, 300, 310,
                        235, 245, 255, 240, 265, 980, 320])
```

```python
# Task 1.1: creating a product dictionary

product = {
    "name": "Notebook",
    "category": "Stationery",
    "price": 1500,
    "quantity": 20,
}

print(product["price"])
```

```python
# Task 1.3: Looping through a dictionary

for product, price in prices.items():
  print(f"{product} cost N{price:,}")

print(prices.get("calculator", "Not stocked"))
```

```python
# Task 1.4: creating discounted prices

sale_prices = {
    product: price * 0.9
    for product, price in prices.items()
}

print(sale_prices)
```

```python
# Task 1.5: counting products in each category

category_count = {}

for item in inventory:
  category = item["category"]

  if category in category_count:
    category_count[category] += 1

  else:
    category_count[category] = 1

print(category_count)
```

```python
# Task 2.1: Calculating total product value

def total_value(price, quantity):
  """calculate the total value of a product"""
  return price * quantity

print(total_value(1500, 40))
```

```python
# Task 2.2: Applying discounts

def apply_discount(price, percent=10):
  """calculate the price after applying a discount"""
  discount = price * (percent / 100)
  return price - discount

print(apply_discount(1500))
print(apply_discount(1500, 25))
```

```python
# Task 2.3: Checking stock status

def stock_status(quantity):
  """Determine the stock status of a product"""

  if quantity == 0:
    return "Out of stock"
  elif quantity < 10:
    return "Low stock"
  else:
    return "In stock"

print(stock_status(0))
print(stock_status(5))
print(stock_status(40))
```

```python
# Task 2.4: Finding products in a category

def products_in_category(inventory, category):
  """Return the naes of products belonging to a category"""

  product_names = []

  for item in inventory:
    if item["category"] == category:
      product_names.append(item["product"])

  return product_names

print(products_in_category(inventory, "Electronics"))

```

```python
# Task 3.1: Checking the shape and data type

print(daily_sales.shape)
print(daily_sales.dtype)
```

```python
# Task 3.2: Calculating daily sales statistics

print("Total sales:", np.sum(daily_sales))
print("Mean sales:", round(np.mean(daily_sales), 2))
print("Maximum sales:", np.max(daily_sales))
print("Standard deviation:", round(np.std(daily_sales), 2))
```

```python
# Task 3.3: Reshaping daily sales into two weeks

weekly_sales = daily_sales.reshape(2, 7)
print(weekly_sales)
print(weekly_sales.shape)
```

```python
from numpy._core.numeric import mean
# Task 3.4: Calculate the median

print("Median sales:", np.median(daily_sales))
print("Mean sales:", np.mean(daily_sales))
```
